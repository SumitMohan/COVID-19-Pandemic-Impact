## 2. Basic DataFrame Structure
These functions help understand the structure and summary of the DataFrame.

```python
# df.head(n): View first n rows (default n=5)
print('\nFirst 3 rows of the dataset:')
print(df.head(3))
# Explanation: Shows the top 3 rows to get a quick look at the data.

# df.tail(n): View last n rows
print('\nLast 2 rows of the dataset:')
print(df.tail(2))
# Explanation: Displays the last 2 rows, useful for checking the end of the dataset.

# df.shape: Get number of rows and columns
print('\nShape of DataFrame:', df.shape)
# Explanation: Returns a tuple (rows, columns), e.g., (10, 12).

# df.columns: Get column names
print('\nColumn names:', df.columns.tolist())
# Explanation: Lists all column names as a list.

# df.index: Get row index
print('\nIndex:', df.index)
# Explanation: Shows the index (row labels) of the DataFrame.

# df.info(): Overview of data types and non-null counts
print('\nDataFrame Info:')
df.info()
# Explanation: Provides a summary of data types, non-null counts, and memory usage.

# df.describe(): Summary statistics for numerical columns
print('\nSummary Statistics:')
print(df.describe())
# Explanation: Shows count, mean, std, min, max, and quartiles for numerical columns.

# df.dtypes: Check data types of each column
print('\nData Types:')
print(df.dtypes)
# Explanation: Lists the data type of each column (e.g., int64, float64, object).

# df.values: Get data as a NumPy array
print('\nData as NumPy array (first 2 rows):')
print(df.values[:2])
# Explanation: Converts the DataFrame to a NumPy array, useful for numerical computations.
```

## 3. Data Selection and Filtering
These functions allow selecting specific rows, columns, or subsets of data.

```python
# df['col']: Select a single column column
print('\nMath Scores:')
print(df['Math_Score'])
# Explanation: Returns a Series containing values of the 'Math_Score' column.

# df[['col1', 'col2']]: Select multiple columns
print('\nSelected Columns (Name, Age):')
print(df[['Name', 'Age']])
# Explanation: Returns a DataFrame with only the specified columns.

# df.iloc[row_idx, col_idx]: Select by position
print('\nRow 1, Columns 0-2 (using iloc):')
print(df.iloc[1, 0:2])
# Explanation: Selects data by integer-based positions (row 1, columns 0 to 1).

# df.loc[row_label, col_label]: Select by label
print('\nRow 0, Columns Name and Math_Score (using loc):')
print(df.loc[0, ['Name', 'Math_Score']])
# Explanation: Selects data by label-based indexing.

# df[df['col'] > value]: Filter rows by condition
print('\nStudents with Math_Score > 85:')
print(df[df['Math_Score'] > 85])
# Explanation: Filters rows where Math_Score is greater than 85.

# df.query('col > value'): SQL-like row filtering
print('\nStudents with Age > 20 (using query):')
print(df.query('Age > 20'))
# Explanation: Filters rows using a SQL-like syntax for conditions.
```

## 4. Data Cleaning
These functions handle missing values, duplicates, and data corrections.

```python
# df.isnull(): Identify missing values
print('\nMissing Values:')
print(df.isnull())
# Explanation: Returns a DataFrame of booleans (True for missing values).

# df.notnull(): Identify non-missing values
print('\nNon-Missing Values:')
print(df.notnull())
# Explanation: Returns True for non-missing values.

# df.dropna(): Drop rows with missing values
print('\nDataFrame after dropping rows with missing values:')
print(df.dropna())
# Explanation: Removes rows with any missing values.

# df.fillna(value): Fill missing values
print('\nDataFrame after filling missing values with 0:')
print(df.fillna(0))
# Explanation: Replaces missing values with the specified value (e.g., 0).

# df.duplicated(): Find duplicate rows
print('\nDuplicate Rows:')
print(df.duplicated())
# Explanation: Returns True for duplicate rows (e.g., same Name and Email).

# df.drop_duplicates(): Remove duplicate rows
print('\nDataFrame after removing duplicates:')
print(df.drop_duplicates())
# Explanation: Drops duplicate rows, keeping the first occurrence.

# df.replace(old, new): Replace specific values
print("\nReplace 'CSE' with 'Computer Science':")
print(df.replace('CSE', 'Computer Science'))
# Explanation: Replaces all occurrences of 'CSE' with 'Computer Science'.

# df.rename(columns={...}): Rename columns
print("\nRename 'Math_Score' to 'Mathematics':")
print(df.rename(columns={'Math_Score': 'Mathematics'}))
# Explanation: Renames specified columns.
```

## 5. Data Transformation
These functions modify or transform data for analysis.

```python
# df.astype(type): Change data type
print('\nConvert Age to float:')
print(df['Age'].astype(float))
# Explanation: Converts the 'Age' column to float type.

# df.apply(func): Apply a function to rows/columns
print('\nAdd 5 to Math_Score using apply:')
print(df['Math_Score'].apply(lambda x: x + 5 if pd.notnull(x) else x))
# Explanation: Applies a function (add 5) to each non-null value in Math_Score.

# df['col'].str.lower(): String operations
print('\nConvert Gender to lowercase:')
print(df['Gender'].str.lower())
# Explanation: Converts all strings in the Gender column to lowercase.

# df['col'].str.lower(): String operations
print('\nConvert Name to lowercase:')
print(df['Name'].str.lower())
# Explanation: Converts all strings in the Name column to lowercase.

# df['col'].str.contains('value'): String matching
print("\nEmails containing 'sharma':")
print(df['Email'].str.contains('sharma', na=False))
# Explanation: Checks if 'sharma' is in each email, returns True/False, handles NaN with na=False.

# pd.to_datetime(df['date']): Convert to datetime
# Adding a date column for demonstration
df['Admission_Date'] = ['2023-06-01', '2023-06-02', '2023-06-01', '2022-06-01', '2023-06-03',
                        '2023-06-04', '2022-06-02', '2023-06-05', '2022-06-03', '2023-06-01']
print('\nConvert Admission_Date to datetime:')
print(pd.to_datetime(df['Admission_Date']))
# Explanation: Converts the Admission_Date column to datetime format.
```

## 6. Aggregation and Grouping
These functions summarize data by grouping and aggregating.

```python
# df.groupby('col'): Group by a column
print('\nAverage Math_Score by Gender:')
print(df.groupby('Gender')['Math_Score'].mean())
# Explanation: Groups data by Gender and computes the mean Math_Score.

# .agg({'col': 'mean'}): Aggregate with a function
print('\nAggregate Math_Score (mean, max):')
print(df.groupby('Gender').agg({'Math_Score': ['mean', 'max']}))
# Explanation: Computes multiple aggregations (mean, max) for Math_Score by Gender.

# .mean(), .sum(), .count(): Common aggregates
print('\nCount of students by Semester:')
print(df.groupby('Semester')['Student_ID'].count())
# Explanation: Counts the number of students in each semester.

# df.pivot_table(): Create pivot tables
print('\nPivot Table: Average Programming_Score by Gender and Semester:')
print(df.pivot_table(values='Programming_Score', index='Gender', columns='Semester', aggfunc='mean'))
# Explanation: Creates a pivot table with mean Programming_Score for each Gender and Semester.
```

## 7. Sorting and Ranking
These functions organize data by sorting or ranking.

```python
# df.sort_values('col'): Sort by a column
print('\nSort by Math_Score (descending):')
print(df.sort_values('Math_Score', ascending=False))
# Explanation: Sorts the DataFrame by Math_Score in descending order.

# df.sort_values(['col1', 'col2']): Multi-column sort
print('\nSort by Semester and Math_Score:')
print(df.sort_values(['Semester', 'Math_Score']))
# Explanation: Sorts by Semester first, then Math_Score.

# df.sort_index(): Sort by index
print('\nSort by index:')
print(df.sort_index())
# Explanation: Sorts the DataFrame by its index.

# df.rank(): Rank values
print('\nRank Math_Score:')
print(df['Math_Score'].rank())
# Explanation: Assigns ranks to Math_Score values (handles ties appropriately).
```

## 8. Column Operations
These functions create, modify, or delete columns.

```python
# df['new'] = df['col1'] + df['col2']: Create a new column
df['Total_Score'] = df['Math_Score'] + df['Programming_Score']
print('\nDataFrame with Total_Score column:')
print(df[['Student_ID', 'Math_Score', 'Programming_Score', 'Total_Score']])
# Explanation: Creates a new column by adding Math_Score and Programming_Score.

# df.drop('col', axis=1): Drop a column
print('\nDrop Total_Score column:')
print(df.drop('Total_Score', axis=1))
# Explanation: Removes the Total_Score column.

# df.insert(loc, 'col', data): Insert a new column at a specific location
df.insert(2, 'First_Name', df['Name'].str.split().str[0])
print('\nDataFrame with First_Name column inserted:')
print(df[['Student_ID', 'Name', 'First_Name']])
# Explanation: Inserts First_Name column at position 2, extracting first names from Name.
```

## 9. Merging, Joining, Concatenating
These functions combine multiple DataFrames.

```python
# Create a second DataFrame for demonstration
df_extra = pd.DataFrame({
    'Student_ID': ['CSE001', 'CSE002', 'CSE011'],
    'Project_Score': [95, 88, 92]
})

# pd.concat([df1, df2]): Stack DataFrames vertically
print('\nConcatenate DataFrames vertically (subset):')
print(pd.concat([df[['Student_ID', 'Name']].head(2), df[['Student_ID', 'Name']].tail(2)]))
# Explanation: Combines rows from two DataFrames vertically.

# df1.merge(df2, on='key'): SQL-style join
print('\nMerge with extra DataFrame on Student_ID:')
print(df.merge(df_extra, on='Student_ID', how='left'))
# Explanation: Merges df with df_extra on Student_ID, keeping all rows from df (left join).

# df1.join(df2): Join on index
# Reset index for join demonstration
df_extra_indexed = df_extra.set_index('Student_ID')
df_indexed = df.set_index('Student_ID').head(3)
print('\nJoin on index:')
print(df_indexed.join(df_extra_indexed, how='left'))
# Explanation: Joins DataFrames on their indices.
```

## 10. Value Counts and Crosstab
These functions analyze frequency and relationships.

```python
# df['col'].value_counts(): Frequency count
print('\nCount of students by Gender:')
print(df['Gender'].value_counts())
# Explanation: Counts the occurrences of each unique value in Gender.

# pd.crosstab(df['col1'], df['col2']): Cross-tabulation
print('\nCrosstab of Gender and Semester:')
print(pd.crosstab(df['Gender'], df['Semester']))
# Explanation: Creates a table showing the frequency of Gender vs. Semester combinations.

# df['col'].unique(): Unique values
print('\nUnique Graduation Years:')
print(df['Graduation_Year'].unique())
# Explanation: Lists all unique values in Graduation_Year.

# df['col'].nunique(): Count of unique values
print('\nNumber of unique Graduation Years:')
print(df['Graduation_Year'].nunique())
# Explanation: Counts the number of unique values in Graduation_Year.
```

## 11. Correlation and Statistics
These functions compute statistical measures.

```python
# df.corr(): Correlation matrix
print('\nCorrelation Matrix:')
print(df[['Math_Score', 'Programming_Score', 'Attendance_Percentage']].corr())
# Explanation: Computes Pearson correlation coefficients between numerical columns.

# df.cov(): Covariance matrix
print('\nCovariance Matrix:')
print(df[['Math_Score', 'Programming_Score', 'Attendance_Percentage']].cov())
# Explanation: Computes the covariance between numerical columns.

# df.std(), df.var(), df.mean(): Descriptive stats
print('\nStandard Deviation of Math_Score:', df['Math_Score'].std())
print('Variance of Programming_Score:', df['Programming_Score'].var())
print('Mean Attendance_Percentage:', df['Attendance_Percentage'].mean())
# Explanation: Computes standard deviation, variance, and mean for specified columns.
```

## 12. Indexing and Resetting
These functions manage the DataFrame's index.

```python
# df.set_index('col'): Set a column as index
print('\nSet Student_ID as index:')
print(df.set_index('Student_ID').head(3))
# Explanation: Sets Student_ID as the index of the DataFrame.

# df.reset_index(): Reset to default integer index
df_indexed = df.set_index('Student_ID')
print('\nReset index:')
print(df_indexed.reset_index().head(3))
# Explanation: Resets the index to default (0, 1, 2, ...).
```

These functions provide additional insights or utilities.

```python
# df.memory_usage(): Memory used by each column
print('\nMemory Usage by Column:')
print(df.memory_usage())
# Explanation: Shows the memory usage (in bytes) for each column.

# df.sample(n=5): Random sample of rows
print('\nRandom sample of 3 rows:')
print(df.sample(n=3))
# Explanation: Returns a random sample of 3 rows from the DataFrame.

# df.copy(): Copy DataFrame
df_copy = df.copy()
print('\nCopied DataFrame (first 2 rows):')
print(df_copy.head(2))
# Explanation: Creates a deep copy of the DataFrame to avoid modifying the original.

# df.empty: Check if DataFrame is empty
print('\nIs DataFrame empty?', df.empty)
# Explanation: Returns True if the DataFrame is empty, False otherwise.
```
